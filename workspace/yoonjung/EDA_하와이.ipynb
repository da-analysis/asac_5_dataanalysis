### time - 리뷰작성날짜

```
%sql

-- 리뷰 작성년도 확인 
-- -> 2019년도가 가장 많음

SELECT 
    YEAR(time) AS year, -- time 컬럼에서 년도를 추출하여 year 컬럼 생성
    COUNT(*) AS review_count -- 년도별 리뷰 개수를 카운트
FROM 
    hive_metastore.silver.review_data
WHERE 
    region = 'Hawaii'
GROUP BY 
    YEAR(time) -- year 컬럼을 기준으로 그룹화
ORDER BY 
    year; -- 년도별로 정렬
```

```
%sql

-- 월별 분포 확인 

SELECT 
    MONTH(time) AS month, -- time 컬럼에서 월을 추출하여 month 컬럼 생성
    COUNT(*) AS review_count -- 월별 리뷰 개수를 카운트
FROM 
    hive_metastore.silver.review_data
WHERE 
    region = 'Hawaii'
GROUP BY 
    MONTH(time) -- month 컬럼을 기준으로 그룹화
ORDER BY 
    month; -- 월별로 정렬
```

### pics - 리뷰 이미지 

```
%sql
-- 리뷰사진 null 값 확인 --> 거의 90% 정도가 리뷰사진 없음

SELECT 
    COUNT(*) AS total_count, -- 전체 개수
    COUNT(CASE WHEN pics IS NULL THEN 1 END) AS null_count, -- pics가 NULL인 경우의 개수
    COUNT(CASE WHEN pics IS NOT NULL THEN 1 END) AS not_null_count -- pics가 NULL이 아닌 경우의 개수
FROM 
    asacdataanalysis.default.review_hawaii;
```

```
%sql
-- 리뷰별 사진 올린 개수 (pics null 값 제외)
-- -> 리뷰 사진 50개까지 업로드한 유저가 있음
SELECT user_id, gmap_id, size(pics) as pics_count
FROM hive_metastore.silver.review_data
WHERE region = 'Hawaii' AND pics IS NOT NULL
order by pics_count desc;
```

## user_id 

```
%sql
--  User_id 기준 - 한 사용자가 몇 개의 gmap_id 장소를 갔는지 (분포 확인)

select min(unique_gmap_count) as min, max(unique_gmap_count) as max, avg(unique_gmap_count) as avg 
from (select user_id, count(distinct gmap_id) as unique_gmap_count
  from hive_metastore.silver.review_data
  where region = 'Hawaii'and user_id is not null
  group by user_id);
```

```
%sql
--  User_id 기준 - 한 사용자가 몇 개의 gmap_id 장소를 갔는지
-- 장소 방문 개수별 분포 확인 --> 1개 장소 방문이 가장 많음

select unique_gmap_count, count(*) as cnt
from 
  (select user_id, count(distinct gmap_id) as unique_gmap_count
  from hive_metastore.silver.review_data
  where region = 'Hawaii' and user_id is not null
  group by user_id)
group by unique_gmap_count
order by unique_gmap_count;
```

```
%sql
-- User_id 기준 - 한 장소에 몇번 리뷰를 작성했는지 
-- 같은 리뷰가 있는 경우가 있어서 중복 건은 합쳐서 카운팅 하고자함 
-- --> 2번 리뷰작성 건 (6건) 제외 시, 대부분 1번 리뷰 작성하였음 
select user_id, gmap_id, count(*) as cnt
from (
  select user_id, gmap_id, text, count(*) as cnt
  from asacdataanalysis.default.review_hawaii
  where user_id is not null
  group by user_id, gmap_id, text)
  group by user_id, gmap_id
  order by cnt desc;
```

```
%sql
--  user_id = '111294719593605274593' and gmap_id = '0x7c00117b634ebe1b:0x766c18df4268fba6' 리스트 확인
select *
from asacdataanalysis.default.review_hawaii
where user_id = '111294719593605274593' and gmap_id = '0x7c00117b634ebe1b:0x766c18df4268fba6';

-- ==> 같은 장소이나 실제 2번 방문 후, 2번 리뷰를 작성한 것으로 확인
```

```
%sql
-- user_id가 null값인 name 확인
SELECT *
FROM asacdataanalysis.default.review_hawaii
WHERE user_id is null;
-- => user_id가 null값인 것을 확인하였을 때, 여행플랫폼 외에도 일반 사용자들도 있음을 확인할 수 있음
```

```
%sql
-- user_id가 null값인 name 확인해보기 --> 여행 플랫폼 리뷰들이 많음 
SELECT name, COUNT(*) AS review_count
FROM asacdataanalysis.default.review_hawaii
WHERE user_id is null
GROUP BY name
ORDER BY review_count DESC;
```

```
%sql
SELECT name, user_id, COUNT(*) AS review_count
FROM asacdataanalysis.default.review_hawaii
GROUP BY name, user_id
ORDER BY review_count DESC;
```

## gmap_id

```
%sql
-- gmap_id 기준 리뷰 개수 분포 확인 (unique 미적용시) 
-- 최소 1개, 최대 9998개, 평균 145개 

SELECT min(gmap_count), max(gmap_count), avg(gmap_count)
from 
  (SELECT gmap_id, COUNT(*) as gmap_count
  from asacdataanalysis.default.review_hawaii
  group by gmap_id); 
  
```

```
%sql
--  gmap_id 기준 - 한 장소에 몇 명의 user_id가 방문(리뷰 작성)했는지
-- 리뷰 개수별 분포 확인
-- -> 8명 유저 수를 가진 gmap_id 가 가장 많았음 

select unique_user_count, count(*) as cnt
from 
  (SELECT gmap_id, COUNT(user_id) as user_count, COUNT(DISTINCT user_id) AS unique_user_count
  from asacdataanalysis.default.review_hawaii
  where user_id is not null
  group by gmap_id)
group by unique_user_count
order by cnt desc;
```
